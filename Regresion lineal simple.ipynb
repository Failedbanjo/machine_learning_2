# REGRESIÓN LINEAL

```python
# Importamos las librerías
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
# Creamos el conjunto de datos
df = pd.DataFrame({
    "horas": [1, 2, 3, 4, 5],
    "nota":  [2, 3, 5, 4, 6]
})
X = df[["horas"]]
y = df[["nota"]]
print(X)
```

```python
# Creamos nuestro modelo
modelo = LinearRegression()
modelo.fit(X, y)
```

```python
# Imprimimos el intercepto y el coeficiente
print("Intercepto:", modelo.intercept_)
print("Pendiente:", modelo.coef_[0])
```

```python
# Realizamos la nueva predicción (con DataFrame para evitar el UserWarning)
nuevo = pd.DataFrame({"horas": [6]})
prediccion = modelo.predict(nuevo)
print("Predicción:", prediccion)
```

```python
# Creamos los puntos de la recta
X_linea = np.linspace(1, 6, 100)
X_linea_df = pd.DataFrame({"horas": X_linea})
y_linea = modelo.predict(X_linea_df)

# Gráfica
plt.figure(figsize=(8, 5))
plt.scatter(X, y, color="blue", label="Datos reales")
plt.plot(X_linea, y_linea, color="red", label="Recta de regresión")
plt.scatter(6, prediccion[0][0], color="green", s=100, zorder=5,
            label=f"Predicción (6 h) = {prediccion[0][0]:.2f}")
plt.xlabel("Horas de estudio")
plt.ylabel("Nota")
plt.title("Regresión lineal: horas vs nota")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Verificación: el punto predicho debe estar sobre la recta
y_recta_en_6 = modelo.intercept_[0] + modelo.coef_[0][0] * 6
print("Valor en la recta (b + m*6):", y_recta_en_6)
print("¿Coincide con la predicción?", np.isclose(y_recta_en_6, prediccion[0][0]))
```
