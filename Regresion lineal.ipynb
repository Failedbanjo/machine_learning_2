# Regresión Lineal

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
```

```python
 #Creamos el conjunto de datos
df = pd.DataFrame({
    "horas":[1,2,3,4,5],
    "nota":[2,3,5,4,6]
})
X= df[["horas"]]
y= df[["nota"]]
print (X)
```

```python
#Creamos nuestro modelo
modelo = LinearRegression()
modelo.fit(X, y)
```

```python
#Imprimimos el interceptor y el coef
print("Interceptor: ", modelo.intercept_)
print("Pendiente:", modelo.coef_[0])
```

```python
#Realizamos la nueva predicción
prediccion = modelo.predict([[6]])
print("Predicción:" , prediccion)
```

```python
#Creamos los puntos de la recta
X_linea = np.linspace(1, 6, 100)
X_linea_df = pd.DataFrame({"horas": X_linea})
y_linea = modelo.predict(X_linea_df)
```

```python
# Graficamos los resultados
plt.scatter(df["horas"], df["nota"], color="blue", label="Datos reales", zorder=5)
plt.plot(X_linea, y_linea, color="red", label="Línea de regresión")
plt.scatter(6, prediccion[0][0], color="green", marker="*", s=150, label="Predicción (6 hrs)", zorder=6)

plt.xlabel("Horas")
plt.ylabel("Nota")
plt.title("Regresión Lineal - Horas vs Nota")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```
